```python
#1. Faça um Programa que leia uma lista de 10 números reais
#   e mostre-os na ordem inversa.
lista = []

for x in range(10):
  #captura 10 numeros float(reais)
  numero = float(input('Informe um numero: '))
  lista.append(numero)#adiciona na lista

print(lista)#imprime os valores na ordem de inserção
lista.sort(reverse=True)#Coloca a lista na ordem reversa
print(lista)#imprime a lista na ordem invertida
```

```python
#2. Faça um Programa que leia uma lista com 4 notas,
#   mostre as notas e a média na tela.
lista_notas = []
for x in range(4):
  nota = float(input('Informe a nota: '))
  lista_notas.append(nota)
#sum --> soma os valores de uma lista (apenas lista numerica)
#len --> quantidade de itens dentro de uma lista
media = sum(lista_notas) / len(lista_notas)

print(lista_notas)
print('Media:', media)
```

```python
#3. Faça um Programa que leia uma lista com 10 caracteres,
#   e diga quantas consoantes foram lidas.
#   Imprima as consoantes.
lista_letras = []
lista_consoantes = []

for x in range(10):
  letra = input('Informe uma letra: ')
  lista_letras.append(letra)

  if letra.lower() not in ['a', 'e', 'i', 'o', 'u']:
    lista_consoantes.append(letra)

print('Total consoantes:', len(lista_consoantes))
print(lista_consoantes)
```

```python
#4. Faça um Programa que leia 20 números inteiros e
#   armazene-os em uma lista. Armazene os números pares na
#   lista PAR e os números IMPARES na lista impar.
#   Imprima as três listas.

lista_numeros = []
lista_par = []
lista_impar = []

for x in range(20):
  numero = int(input('Informe um valor: '))
  lista_numeros.append(numero)
  if numero % 2 == 0:
    lista_par.append(numero)
  else:
    lista_impar.append(numero)

print('Numeros:', lista_numeros)
print('Pares:', lista_par)
print('Impares:', lista_impar)
```

```python
#5. Faça um Programa que peça as quatro notas de 10 alunos,
#   calcule e armazene em uma lista a média de cada aluno,
#   imprima o número de alunos com média maior ou igual a 7,0.

lista_alunos = []
lista_media = []
contador_acima_media = 0

for x in range(10):
  #captura as informações de 10 alunos
  aluno = []
  for y in range(4):
    #captura as 4 notas de cada aluno
    nota = float(input('Informe a nota: '))
    aluno.append(nota)#armazena a nota do aluno
  lista_alunos.append(aluno)
  #calcula a media do aluno
  media = sum(aluno)/len(aluno)
  lista_media.append(media)

for x in lista_media:
  if x >= 7.0:
    contador_acima_media = contador_acima_media + 1

print('Alunos:', lista_alunos)
print('Medias:', lista_media)
print('Alunos acima da média:', contador_acima_media)
```

```python
#6. Faça um Programa que leia duas listas com 10 elementos
#   cada. Gere uma terceira lista de 20 elementos, cujos
#   valores deverão ser compostos pelos elementos
#   intercalados das duas outras listas.

#lista_01 --> a, b, c, d, e
#lista_02 --> 1, 2, 3, 4, 5
#lista_03 --> a, 1, b, 2, c, 3, d, 4, e, 5

lista_01 = []
lista_02 = []
lista_03 = []
#povoando a lista_01
print('Insira os dados da lista 01')
for x in range(10):
  valor = input('Insira o valor: ')
  lista_01.append(valor)

#povoando a lista_02
print('Insira os dados da lista 02')
for x in range(10):
  valor = input('Insira o valor: ')
  lista_02.append(valor)

#intercalando as lista
for x in range(10):
  lista_03.append(lista_01[x])
  lista_03.append(lista_02[x])

print('Lista 01 -->', lista_01)
print('Lista 02 -->', lista_02)
print('Lista 03 -->', lista_03)
```

```python
#7.
numeros = [5, 1, 10, 3, 8, 2]

# Use a função len() para imprimir quantos itens existem na
# lista.
print(len(numeros)) #6

# Use o operador in para verificar se o número 10 está
# na lista e imprima o resultado (deve ser True ou False).
print(10 in numeros) #True

# Use a função sorted() para imprimir uma nova lista com
# os números ordenados, sem alterar a lista numeros original.

lista_ordenada = sorted(numeros)
print(lista_ordenada) #[1, 2, 3, 5, 8, 10]

# Imprima a lista numeros original para confirmar que ela
# não mudou.
#Ordena, mas NÃO altera a lista original (cria uma copia)
print(numeros) #[5, 1, 10, 3, 8, 2]

# Agora, use o método sort() para ordenar a lista numeros
# no local (modificando a original).
numeros.sort()

# Imprima a lista numeros novamente para ver
# o resultado da ordenação.
#Ordena e altera a lista original
print(numeros) #[1, 2, 3, 5, 8, 10]
```
